```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Set plot style and load data
sns.set_theme(style="whitegrid")
metrics_df = pd.read_csv("training_metrics.csv")
```

```python
# --- Reward Plots ---
plt.figure(figsize=(12, 6))
plt.plot(metrics_df['iteration'], metrics_df['p0_reward_mean'], label='P0 Mean Reward', color='blue')
plt.plot(metrics_df['iteration'], metrics_df['p1_reward_mean'], label='P1 Mean Reward', color='red')
plt.xlabel('Training Iteration')
plt.ylabel('Mean Reward')
plt.title('Per-Agent Mean Rewards during Training')
plt.legend()
plt.savefig('plots/multi_agent_mean_rewards.pdf')
plt.show()
```

```python
# Total Loss
plt.figure(figsize=(12, 6))
plt.plot(metrics_df['iteration'], metrics_df['p0_total_loss'], label='P0 Total Loss', color='blue', alpha=0.7)
plt.plot(metrics_df['iteration'], metrics_df['p1_total_loss'], label='P1 Total Loss', color='red', alpha=0.7)
plt.xlabel('Training Iteration')
plt.ylabel('Total Loss')
plt.title('Total Loss per Agent')
plt.legend()
plt.savefig('plots/multi_agent_total_loss.pdf')
plt.show()
```

```python
# Policy loss
window_size = 10
metrics_df['p0_policy_loss_avg'] = metrics_df['p0_policy_loss'].rolling(window=window_size, min_periods=1).mean()
metrics_df['p1_policy_loss_avg'] = metrics_df['p1_policy_loss'].rolling(window=window_size, min_periods=1).mean()

plt.figure(figsize=(12, 6))

# Plot original data with reduced alpha
plt.plot(metrics_df['iteration'], metrics_df['p0_policy_loss'], label='P0 Policy Loss (Raw)', color='lightblue', alpha=0.5)
plt.plot(metrics_df['iteration'], metrics_df['p1_policy_loss'], label='P1 Policy Loss (Raw)', color='lightcoral', alpha=0.5)

# Plot rolling average lines
plt.plot(metrics_df['iteration'], metrics_df['p0_policy_loss_avg'], label=f'P0 Policy Loss ({window_size}-iter Avg)', color='blue', linewidth=2)
plt.plot(metrics_df['iteration'], metrics_df['p1_policy_loss_avg'], label=f'P1 Policy Loss ({window_size}-iter Avg)', color='red', linewidth=2)

# Labels and Title
plt.xlabel('Training Iteration')
plt.ylabel('Policy Loss')
plt.title('Policy Loss per Agent (with Moving Average)')
plt.legend()
plt.savefig('plots/multi_agent_policy_loss_avg.pdf')
plt.show()
```

```python
window_size = 10
metrics_df['p0_vf_loss_avg'] = metrics_df['p0_vf_loss'].rolling(window=window_size, min_periods=1).mean()
metrics_df['p1_vf_loss_avg'] = metrics_df['p1_vf_loss'].rolling(window=window_size, min_periods=1).mean()

plt.figure(figsize=(12, 6))

# Plot original data with reduced alpha
plt.plot(metrics_df['iteration'], metrics_df['p0_vf_loss'], label='P0 VF Loss (Raw)', color='lightblue', alpha=0.5)
plt.plot(metrics_df['iteration'], metrics_df['p1_vf_loss'], label='P1 VF Loss (Raw)', color='lightcoral', alpha=0.5)

# Plot rolling average lines
plt.plot(metrics_df['iteration'], metrics_df['p0_vf_loss_avg'], label=f'P0 VF Loss ({window_size}-iter Avg)', color='blue', linewidth=2)
plt.plot(metrics_df['iteration'], metrics_df['p1_vf_loss_avg'], label=f'P1 VF Loss ({window_size}-iter Avg)', color='red', linewidth=2)

# Labels and Title
plt.xlabel('Training Iteration')
plt.ylabel('VF Loss')
plt.title('Value Function Loss per Agent (with Moving Average)')
plt.legend()
plt.savefig('plots/multi_agent_vf_loss_avg.pdf')
plt.show()
```

```python
# Entropy
plt.figure(figsize=(12, 6))
plt.plot(metrics_df['iteration'], metrics_df['p0_entropy'], label='P0 Entropy', color='blue', alpha=0.7)
plt.plot(metrics_df['iteration'], metrics_df['p1_entropy'], label='P1 Entropy', color='red', alpha=0.7)
plt.xlabel('Training Iteration')
plt.ylabel('Entropy')
plt.title('Policy Entropy per Agent')
plt.legend()
plt.savefig('plots/multi_agent_entropy.pdf')
plt.show()
```

```python
# KL Divergence
plt.figure(figsize=(12, 6))
plt.plot(metrics_df['iteration'], metrics_df['p0_kl_loss'], label='P0 KL Divergence', color='blue', alpha=0.7)
plt.plot(metrics_df['iteration'], metrics_df['p1_kl_loss'], label='P1 KL Divergence', color='red', alpha=0.7)
plt.xlabel('Training Iteration')
plt.ylabel('KL Divergence')
plt.title('KL Divergence per Agent')
plt.legend()
plt.savefig('plots/multi_agent_kl_divergence.pdf')
plt.show()
```
